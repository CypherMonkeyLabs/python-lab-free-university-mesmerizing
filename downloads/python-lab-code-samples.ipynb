## Functions, classes, and methods: Decorators

```python
from functools import cache

@cache
def fib(n):
    return n if n < 2 else fib(n - 1) + fib(n - 2)

print(fib(30))
```

## Flow: iterators, generators, and async: Tasks and TaskGroup

Run this one on your own computer. The browser practice desk uses await inside checks instead of asyncio.run.

```python
import asyncio

async def sq(n):
    return n * n

async def main():
    async with asyncio.TaskGroup() as tg:
        ts = [tg.create_task(sq(i)) for i in range(3)]
    print([t.result() for t in ts])

asyncio.run(main())
```

## Guide: Control flow: choose, repeat, stop

```python
def first_even(nums):
    for n in nums:
        if n % 2 == 0:
            return n
    return None

for n in [1, 3, 5]:
    if n > 4:
        print('found', n)
        break
else:
    print('nothing over 4')

match (3, 4):
    case (x, y) if x == y:
        print('same')
    case (x, y):
        print('sum', x + y)
print(first_even([1, 3, 6]))
```

## Guide: Iterators and generators

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

it = countdown(3)
print(next(it))
print(list(it))
print(sum(x * x for x in range(4)))
```

## Guide: Async in practice

Run this one on your own computer. The browser practice desk uses await inside checks instead of asyncio.run.

```python
import asyncio

async def sq(n):
    await asyncio.sleep(0)
    return n * n

async def main():
    async with asyncio.TaskGroup() as tg:
        tasks = [tg.create_task(sq(i)) for i in range(4)]
    print([t.result() for t in tasks])

asyncio.run(main())
```

## Guide: Functions and what a call does

```python
def add_item(item, bucket=None):
    if bucket is None:
        bucket = []
    bucket.append(item)
    return bucket

print(add_item('a'))
print(add_item('b'))

def counter():
    n = 0
    def step():
        nonlocal n
        n += 1
        return n
    return step

c = counter()
c()
print(c())
```

## Guide: Classes, methods, and dataclasses

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

    @classmethod
    def from_text(cls, text):
        a, b = text.split(',')
        return cls(int(a), int(b))

    def total(self):
        return self.x + self.y

p = Point.from_text('3,4')
print(p)
print(p.total(), p == Point(3, 4))
```
